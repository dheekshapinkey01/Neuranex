```python
import pandas as pd

# Load job descriptions
jobs_df = pd.read_csv("../data/raw/job_descriptions.csv")

print("Job descriptions loaded successfully!")
print("Shape:", jobs_df.shape)
```

```python
jobs_df["job_text"] = (
    jobs_df["job_title"].fillna("") + " " +
    jobs_df["job_description"].fillna("") + " " +
    jobs_df["required_skills"].fillna("")
)

print("Job text prepared successfully!")
print("Number of job descriptions:", len(jobs_df))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create a separate vectorizer for job matching
job_vectorizer = TfidfVectorizer(
    stop_words="english"
)

# Convert job descriptions into TF-IDF vectors
job_tfidf_matrix = job_vectorizer.fit_transform(
    jobs_df["job_text"]
)

print("Job TF-IDF transformation completed!")
print("Job TF-IDF matrix shape:", job_tfidf_matrix.shape)
```

```python
import pandas as pd

# Load the cleaned resume dataset
cleaned_df = pd.read_csv("../data/processed/cleaned_resumes.csv")

print("Cleaned resume dataset loaded successfully!")
print("Shape:", cleaned_df.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

sample_resume = cleaned_df["cleaned_resume_text"].iloc[0]

resume_vector = job_vectorizer.transform([sample_resume])

similarity_scores = cosine_similarity(
    resume_vector,
    job_tfidf_matrix
)[0]

print("Similarity calculation completed!")
print("Number of job scores:", len(similarity_scores))
```

```python
# Find the job with the highest similarity score

best_job_index = similarity_scores.argmax()

best_job = jobs_df.iloc[best_job_index]["job_title"]
best_score = similarity_scores[best_job_index]

print("Best Matching Job:", best_job)
print(f"Match Score: {best_score * 100:.2f}%")
```

```python
# Display match scores for all jobs

job_match_results = pd.DataFrame({
    "Job Title": jobs_df["job_title"],
    "Match Score": similarity_scores * 100
})

job_match_results = job_match_results.sort_values(
    by="Match Score",
    ascending=False
).reset_index(drop=True)

job_match_results["Match Score"] = job_match_results["Match Score"].round(2)

job_match_results
```

```python
# Function to compare resume skills with required job skills

def compare_skills(resume_skills, required_skills):
    resume_skills_lower = {
        skill.lower().strip()
        for skill in resume_skills
    }

    required_skills_list = [
        skill.strip()
        for skill in required_skills.split(",")
    ]

    matching_skills = []
    missing_skills = []

    for skill in required_skills_list:
        if skill.lower() in resume_skills_lower:
            matching_skills.append(skill)
        else:
            missing_skills.append(skill)

    return matching_skills, missing_skills


print("Skill comparison function created successfully!")
```

```python
import re

# Skill dictionary
job_skills = {
    "AI/ML Engineer": [
        "Python", "Machine Learning", "Deep Learning",
        "TensorFlow", "PyTorch", "NLP",
        "NumPy", "Pandas", "Scikit-learn"
    ],
    "Data Analyst": [
        "SQL", "Excel", "Power BI", "Tableau",
        "Python", "Data Visualization", "Statistics", "Pandas"
    ],
    "Data Scientist": [
        "Python", "Statistics", "Machine Learning",
        "Pandas", "NumPy", "Scikit-learn",
        "SQL", "Data Visualization"
    ],
    "Software Developer": [
        "Java", "C++", "Python", "Data Structures",
        "Algorithms", "OOP", "Git", "SQL"
    ],
    "Web Developer": [
        "HTML", "CSS", "JavaScript", "React",
        "Node.js", "Git", "REST API", "Web Development"
    ],
    "Cybersecurity Analyst": [
        "Network Security", "Linux", "SIEM",
        "Ethical Hacking", "Cryptography", "Python",
        "Cybersecurity", "Vulnerability Assessment"
    ],
    "Cloud Engineer": [
        "AWS", "Azure", "Cloud Computing", "Linux",
        "Docker", "Networking", "Terraform", "Cloud Security"
    ],
    "Database Administrator": [
        "SQL", "MySQL", "PostgreSQL", "Oracle",
        "Database Management", "Linux",
        "Database Security", "Backup"
    ],
    "DevOps Engineer": [
        "Docker", "Kubernetes", "Jenkins", "Git",
        "CI/CD", "AWS", "Linux", "Terraform"
    ],
    "Mobile App Developer": [
        "Android", "Kotlin", "Java", "Flutter",
        "Dart", "Firebase", "Mobile Development", "Git"
    ],
    "QA/Test Engineer": [
        "Manual Testing", "Selenium", "Java", "Python",
        "JUnit", "Test Automation", "API Testing", "Software Testing"
    ],
    "Business Analyst": [
        "SQL", "Excel", "Power BI", "Requirements Analysis",
        "Documentation", "Agile", "Business Analysis", "Data Analysis"
    ]
}

# Skill extraction function
def extract_skills(text, skill_dictionary):
    text = text.lower()
    extracted_skills = []

    for category_skills in skill_dictionary.values():
        for skill in category_skills:
            pattern = r"\b" + re.escape(skill.lower()) + r"\b"

            if re.search(pattern, text):
                extracted_skills.append(skill)

    return list(dict.fromkeys(extracted_skills))


print("Skill extraction function loaded successfully!")
```

```python
# Get skills from the first resume
resume_skills = extract_skills(
    cleaned_df["cleaned_resume_text"].iloc[0],
    job_skills
)

# Get required skills for the best-matching job
required_skills = jobs_df.iloc[best_job_index]["required_skills"]

# Compare skills
matching_skills, missing_skills = compare_skills(
    resume_skills,
    required_skills
)

print("Best Matching Job:", best_job)

print("\nMatching Skills:")
print(matching_skills)

print("\nMissing Skills:")
print(missing_skills)
```

```python
# Generate skill recommendations from missing skills

def recommend_skills(missing_skills, max_recommendations=5):
    """
    Recommend missing skills that can improve the resume's
    match with the selected job.
    """
    
    recommendations = missing_skills[:max_recommendations]

    return recommendations


recommended_skills = recommend_skills(missing_skills)

print("Recommended Skills:")
print(recommended_skills)
```
