```python
import pandas as pd
import random
from pathlib import Path

random.seed(42)

# Create the raw data folder
data_path = Path("../data/raw")
data_path.mkdir(parents=True, exist_ok=True)

# 12 job categories and their core skills
job_categories = {
    "AI/ML Engineer": [
        "Python", "Machine Learning", "Deep Learning", "TensorFlow",
        "PyTorch", "NLP", "NumPy", "Pandas", "Scikit-learn"
    ],

    "Data Analyst": [
        "SQL", "Excel", "Power BI", "Tableau", "Python",
        "Data Visualization", "Statistics", "Pandas"
    ],

    "Data Scientist": [
        "Python", "Statistics", "Machine Learning", "Pandas",
        "NumPy", "Scikit-learn", "SQL", "Data Visualization"
    ],

    "Software Developer": [
        "Java", "C++", "Python", "Data Structures", "Algorithms",
        "OOP", "Git", "SQL"
    ],

    "Web Developer": [
        "HTML", "CSS", "JavaScript", "React", "Node.js",
        "Git", "REST API", "Web Development"
    ],

    "Cybersecurity Analyst": [
        "Network Security", "Linux", "SIEM", "Ethical Hacking",
        "Cryptography", "Python", "Cybersecurity", "Vulnerability Assessment"
    ],

    "Cloud Engineer": [
        "AWS", "Azure", "Cloud Computing", "Linux",
        "Docker", "Networking", "Terraform", "Cloud Security"
    ],

    "Database Administrator": [
        "SQL", "MySQL", "PostgreSQL", "Oracle",
        "Database Management", "Linux", "Database Security", "Backup"
    ],

    "DevOps Engineer": [
        "Docker", "Kubernetes", "Jenkins", "Git",
        "CI/CD", "AWS", "Linux", "Terraform"
    ],

    "Mobile App Developer": [
        "Android", "Kotlin", "Java", "Flutter",
        "Dart", "Firebase", "Mobile Development", "Git"
    ],

    "QA/Test Engineer": [
        "Manual Testing", "Selenium", "Java", "Python",
        "JUnit", "Test Automation", "API Testing", "Software Testing"
    ],

    "Business Analyst": [
        "SQL", "Excel", "Power BI", "Requirements Analysis",
        "Documentation", "Agile", "Business Analysis", "Data Analysis"
    ]
}

print("Number of job categories:", len(job_categories))
print("\nCategories:")
for category in job_categories:
    print("-", category)
```

```python
# Generate synthetic resume records

resume_records = []

resume_id = 1

for category, skills in job_categories.items():
    for i in range(100):

        # Select some skills from the category
        selected_skills = random.sample(
            skills,
            k=random.randint(5, len(skills))
        )

        # Add a few common skills
        common_skills = [
            "Communication",
            "Problem Solving",
            "Teamwork",
            "Git",
            "Project Management"
        ]

        extra_skills = random.sample(
            common_skills,
            k=random.randint(1, 2)
        )

        all_skills = selected_skills + extra_skills

        # Create resume text
        resume_text = (
            f"Experienced professional interested in {category}. "
            f"Skills include {', '.join(all_skills)}. "
            f"Worked on academic and practical projects involving "
            f"{', '.join(selected_skills[:4])}. "
            f"Familiar with software development practices, "
            f"problem solving and teamwork."
        )

        resume_records.append({
            "resume_id": f"R{resume_id:04d}",
            "resume_text": resume_text,
            "job_category": category
        })

        resume_id += 1

# Convert the records into a DataFrame
resumes_df = pd.DataFrame(resume_records)

# Save the dataset
resume_file = data_path / "resumes.csv"
resumes_df.to_csv(resume_file, index=False)

print("Resume dataset created successfully!")
print("Number of resumes:", len(resumes_df))
print("Number of columns:", len(resumes_df.columns))
print("Dataset saved to:", resume_file)
```
